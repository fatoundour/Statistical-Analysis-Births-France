# Analyse universitaire originale — naissances en France (2022)
Ce notebook conserve la méthode du projet initial : chaque moyenne journalière mensuelle est répétée sur les jours du mois pour la visualisation. **Les 365 valeurs ne sont pas des observations journalières indépendantes.** La statistique et la p-value historiques sont conservées à titre de reproduction pédagogique ; voir `scripts/test_mensuel.py` pour un test fondé sur les 12 mois.


```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Charger le fichier CSV préparé à partir des données originales
df = pd.read_csv("data/naissances_moyennes_journalieres_2022.csv")

# Extraire les moyennes journalières par mois en 2022
moyennes_mois = df["moyenne_journaliere_2022"].values  # Moyennes journalières mensuelles publiées

# Nombre de jours par mois
jours_par_mois = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

# Étendre les moyennes mensuelles sur 365 jours
naissances_par_jour = []
for i in range(12):  # Boucle sur les mois (0 à 11)
    for j in range(jours_par_mois[i]):  # Boucle sur les jours du mois
        naissances_par_jour.append(moyennes_mois[i])  # Remplir avec la moyenne du mois

# Créer un DataFrame avec les 365 valeurs
df_final = pd.DataFrame(naissances_par_jour, columns=["Naissances Moyennes"])

# Valeurs observées (nombre de naissances par jour)
observed = df_final["Naissances Moyennes"].values

# Valeurs attendues sous H0 (répartition uniforme des naissances)
moyenne_totale = observed.sum()  # Somme des naissances sur les 365 jours
expected = [moyenne_totale / 365] * 365  # Répartition uniforme sur 365 jours

# Test du chi-deux
chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)

# Affichage des résultats
print(f"Statistique du chi-2 : {chi2_stat:.2f}")
print(f"P-valeur : {p_value:.5f}")

# Visualisation des résultats
plt.figure(figsize=(4, 4))

# Plot des valeurs observées (ligne bleue)
plt.plot(range(365), observed, label="Observé", color='blue', linestyle='-', linewidth=2)

# Plot des valeurs attendues (ligne rouge)
plt.plot(range(365), expected, label="Attendu (Uniforme)", color='red', linestyle='--', linewidth=2)

# Ajouter des labels et une légende
plt.title("Comparaison entre les naissances observées et attendues")
plt.xlabel("Jour de l'année")
plt.ylabel("Nombre de naissances")
plt.legend()

# Afficher le graphique
plt.tight_layout()
plt.show()

# Interprétation
alpha = 0.05
if p_value < alpha:
    print("Rejet de H0 : La répartition des naissances n'est PAS uniforme.")
else:
    print("On ne rejette pas H0 : Les naissances pourraient être uniformément réparties.")
```
